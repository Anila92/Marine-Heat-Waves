```python
import numpy as np
from datetime import date
from matplotlib import pyplot as plt
import xarray as xr
```

```python
mhw_dict_ob = np.load('path/OBS_mhw_dict_1982_2022.npy', allow_pickle=True).item()
mhw_dict_dcps = np.load('path/IITM_DCPS_mhw_dict_ens_lead1.npy', allow_pickle=True).item()
mhw_dict_dcps_corr = np.load('path/bias_corrected_quantile_IITM_DCPS_mhw_dict_ens_lead1_1982_2022.npy', allow_pickle=True).item()
```

```python
ds = xr.open_dataset('path/OISST_daily_mean_1982_2022.nc')
sst_ob = ds.sst.sel(time=slice('1983', '1984'), lat=slice(-10,30), lon=slice(40,120))
ds = xr.open_dataset('path/IITM_DCPS_sst_day_1982_2022_ensmean_LY1_rg1by4.nc')
sst_dcps = ds.sst.sel(time=slice('1983', '1984'), lat=slice(-10,30), lon=slice(40,119.75))
ds = xr.open_dataset('path/bias_corrected_quantile_IITM_DCPS_daily_sst_1982_2022.nc')
sst_dcps_bias = ds.sst.sel(time=slice('1982', '2022'),lat=slice(-10,30), lon=slice(40,120))

sst_ob.shape
```

```python
n_events_ob = xr.DataArray(np.zeros((len(sst_ob.lat), len(sst_ob.lon))), coords=(sst_ob.lat, sst_ob.lon), dims=['lat','lon'])
n_days_ob = xr.DataArray(np.zeros((len(sst_ob.lat), len(sst_ob.lon))), coords=(sst_ob.lat, sst_ob.lon), dims=['lat','lon'])

for i in range(len(sst_ob.lat)):
    for j in range(len(sst_ob.lon)):
        for k in range(mhw_dict_ob[(i,j)][0]['n_events']):
            start_date = mhw_dict_ob[(i,j)][0]['date_start'][k]
            end_date = mhw_dict_ob[(i,j)][0]['date_end'][k]
            if 1984 < start_date.year < 2017 and 1984 < end_date.year < 2017:
                n_events_ob[i,j] += 1
                n_days_ob[i,j] += mhw_dict_ob[(i,j)][0]['duration'][k]

n_events_ob.shape
```

```python
n_events_ob.to_netcdf('path/n_events_OBS_1984_2017.nc', 'w')
n_days_ob.to_netcdf('path/n_days_OBS_1984_2017.nc', 'w')
```

```python
n_events_dcps = xr.DataArray(np.zeros((len(sst_dcps.lat), len(sst_dcps.lon))), coords=(sst_dcps.lat, sst_dcps.lon), dims=['lat','lon'])
n_days_dcps = xr.DataArray(np.zeros((len(sst_dcps.lat), len(sst_dcps.lon))), coords=(sst_dcps.lat, sst_dcps.lon), dims=['lat','lon'])

for i in range(len(sst_dcps.lat)):
    for j in range(len(sst_dcps.lon)):
        for k in range(mhw_dict_dcps[(i,j)][0]['n_events']):
            start_date = mhw_dict_dcps[(i,j)][0]['date_start'][k]
            end_date = mhw_dict_dcps[(i,j)][0]['date_end'][k]
            if 1984 < start_date.year < 2017 and 1984 < end_date.year < 2017:
                n_events_dcps[i,j] += 1
                n_days_dcps[i,j] += mhw_dict_dcps[(i,j)][0]['duration'][k]

n_events_dcps.shape
```

```python
n_events_dcps.to_netcdf('path/n_events_IITM_DCPS_1984_2017.nc', 'w')
n_days_dcps.to_netcdf('path/n_days_IITM_DCPS_1984_2017.nc', 'w')
```
